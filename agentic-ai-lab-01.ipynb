```python
# Extended Rule Based Weather Advisor: Task 01

def weather_advice(temperature_c, humidity_percent, wind_speed_kmh, air_quality_index):
    advice = []

    # Temperature rules
    if temperature_c >= 50:
        advice.append("Extreme heat: stay hydrated and avoid the sun.")
    elif temperature_c <= -10:
        advice.append("Cold weather: wear a warm jacket.")

    # Humidity rule
    if humidity_percent >= 80:
        advice.append("High humidity: expect possible rain, carry an umbrella.")

    # New Rule 1: Wind speed
    if wind_speed_kmh >= 50:
        advice.append("Strong winds: avoid outdoor activities and secure loose objects.")

    # New Rule 2: Air quality
    if air_quality_index >= 150:
        advice.append("Poor air quality: avoid prolonged outdoor activities and consider wearing a mask.")
    elif air_quality_index >= 100:
        advice.append("Moderate air quality: sensitive people should limit prolonged outdoor activities.")

    # Default advice
    if not advice:
        advice.append("Conditions look moderate. No special precautions needed.")

    return advice


if __name__ == "__main__":
    # Six test cases
    test_cases = [
        (52, 40, 20, 50),
        (-13, 55, 10, 40),
        (32, 85, 15, 70),
        (50, 50, 55, 80),
        (25, 60, 65, 160),
        (22, 45, 15, 110)
    ]

    print("WEATHER ADVISORY TEST RESULTS")
    print("=" * 70)

    for i, (temp, humidity, wind, aqi) in enumerate(test_cases, start=1):
        result = weather_advice(temp, humidity, wind, aqi)

        print(f"\nTest Case {i}:")
        print(f"Temperature   : {temp}°C")
        print(f"Humidity      : {humidity}%")
        print(f"Wind Speed    : {wind} km/h")
        print(f"Air Quality   : AQI {aqi}")
        print("Advice        :")

        for message in result:
            print(f"  - {message}")
```

```python
# Simple Keyword-Based Spam Classifier: TASK 02

# List of words/phrases commonly associated with spam emails
spam_keywords = [
    "free",
    "winner",
    "urgent",
    "click here",
    "prize"
]

# If an email contains 2 or more indicators, classify it as spam.
THRESHOLD = 2


def classify_email(email_text):
    """Classify an email as Spam or Legitimate using keyword counting."""
    text = email_text.lower()

    # Count how many spam keywords appear in the email
    keyword_count = 0

    for keyword in spam_keywords:
        if keyword in text:
            keyword_count += 1

    if keyword_count >= THRESHOLD:
        label = "Spam"
    else:
        label = "Legitimate"

    return label, keyword_count


# In principle, this is like facial recognition or medical image diagnosis:
# each system assigns an input to one category from a fixed set of categories.


# Six sample emails with their correct labels
test_emails = [
    {
        "text": "Congratulations! You are a winner. Click here to claim your prize!",
        "correct_label": "Spam"
    },
    {
        "text": "URGENT! You have won a free prize. Click here immediately!",
        "correct_label": "Spam"
    },
    {
        "text": "Hi Sarah, the meeting has been moved to 3 PM tomorrow. Please confirm.",
        "correct_label": "Legitimate"
    },
    {
        "text": "Your order has been shipped and will arrive on Friday. Thank you for shopping with us.",
        "correct_label": "Legitimate"
    },
    {
        "text": "You are a winner! Your free gift is waiting for you. Click here to receive it.",
        "correct_label": "Spam"
    },
    {
        "text": "Reminder: your electricity bill is due on September 20. Please make your payment online.",
        "correct_label": "Legitimate"
    }
]


# Test the classifier
correct_predictions = 0

print("=" * 70)
print("KEYWORD-BASED EMAIL SPAM CLASSIFIER")
print("=" * 70)

for i, email in enumerate(test_emails, start=1):
    predicted_label, keyword_count = classify_email(email["text"])

    if predicted_label == email["correct_label"]:
        correct_predictions += 1
        result = "Correct"
    else:
        result = "Incorrect"

    print(f"\nEmail {i}")
    print("-" * 70)
    print(f"Email text      : {email['text']}")
    print(f"Spam keywords   : {keyword_count}")
    print(f"Correct label   : {email['correct_label']}")
    print(f"Classifier      : {predicted_label}")
    print(f"Result          : {result}")


# Calculate accuracy
total_emails = len(test_emails)
accuracy = (correct_predictions / total_emails) * 100

print("\n" + "=" * 70)
print("CLASSIFIER PERFORMANCE")
print("=" * 70)
print(f"Correct predictions : {correct_predictions}/{total_emails}")
print(f"Accuracy            : {accuracy:.2f}%")
```

```python
# Task 2 Extension: Testing Different Spam Keyword Thresholds Task 3

spam_keywords = [
    "free",
    "winner",
    "urgent",
    "click here",
    "prize"
]


def classify_email(email_text, threshold):
    """Classify an email based on the number of spam keywords."""
    text = email_text.lower()

    keyword_count = 0

    for keyword in spam_keywords:
        if keyword in text:
            keyword_count += 1

    if keyword_count >= threshold:
        return "Spam"
    else:
        return "Legitimate"


# The same six sample emails from Task 2
test_emails = [
    {
        "text": "Congratulations! You are a winner. Click here to claim your prize!",
        "correct_label": "Spam"
    },
    {
        "text": "URGENT! You have won a free prize. Click here immediately!",
        "correct_label": "Spam"
    },
    {
        "text": "Hi Sarah, the meeting has been moved to 3 PM tomorrow. Please confirm.",
        "correct_label": "Legitimate"
    },
    {
        "text": "Your order has been shipped and will arrive on Friday. Thank you for shopping with us.",
        "correct_label": "Legitimate"
    },
    {
        "text": "You are a winner! Your free gift is waiting for you. Click here to receive it.",
        "correct_label": "Spam"
    },
    {
        "text": "Reminder: your electricity bill is due on September 20. Please make your payment online.",
        "correct_label": "Legitimate"
    }
]


# Test thresholds from 1 to 5
print("=" * 60)
print("SPAM CLASSIFIER THRESHOLD COMPARISON")
print("=" * 60)

best_threshold = None
best_accuracy = -1

for threshold in range(1, 6):
    correct_predictions = 0

    for email in test_emails:
        prediction = classify_email(email["text"], threshold)

        if prediction == email["correct_label"]:
            correct_predictions += 1

    accuracy = (correct_predictions / len(test_emails)) * 100

    print(
        f"Threshold = {threshold}: "
        f"{correct_predictions}/6 correct, "
        f"Accuracy = {accuracy:.2f}%"
    )

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_threshold = threshold


print("\n" + "=" * 60)
print(f"Best threshold: {best_threshold}")
print(f"Best accuracy : {best_accuracy:.2f}%")
print("=" * 60)
```
